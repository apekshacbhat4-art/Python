**1)Coin toss game!**

```python
import random
coin_sides=["H","T"]
count,won,lost=0,0,0
print("\t\t***Welcome to coin toss game!!***\n")
continue_choice=1
while (continue_choice):
  count+=1
  print(f"\n\n\t\t\t\tROUND {count}")
  print("_"*100)
  choice=input("\n\nWhat do you choose ? Heads or Tails ?\n\nEnter H for Heads and T for Tails : ")
  if (random.choice(coin_sides)==choice):
    print(f"\n\nYou won!")
    won+=1
  else :
    print(f"\n\nYou lost!")
    lost+=1
  continue_choice=input("\n\nDo you wanna try again ? (yes/no):")
  if(continue_choice.lower()=="yes"):
    continue_choice=1
  elif (continue_choice.lower()=="no"):
    continue_choice=0
  else :
    print("Error in choice")
    continue_choice=0
print("\n")
print("*"*15+"MATCH SUMMARY"+"*"*15)
print(f"Total rounds played     : {count}")
print(f"You won                 : {won}")
print(f"You lost                : {lost}")
print(f"Winning percentage      : {int((won/count)*100)}")
print("*"*43)
print("\nThank you for playing!!")
```

**4) Password checker**

```python
def islower(user_input) :
  for i in user_input :
    if(ord(i)>=97 and ord(i)<=122) :
      return True
  return False
def isupper(user_input) :
  for i in user_input :
    if (ord(i)>=65 and ord(i)<=90) :
      return True
  return False
def isspecial_char(user_input) :
  for i in user_input :
    str="@#$!%&*"
    if ( i in str):
      return True
  return False
def is_digit(user_input) :
  for i in user_input :
    if(i.isdigit()==True) :
      return True
  return False
def password_strength(user_input) :
  sucess=0
  if (len(user_input)>=8) :
    sucess+=1
  if (islower(user_input)==True) :
    sucess+=1
  if (isupper(user_input)==True) :
    sucess+=1
  if(is_digit(user_input)==True) :
    sucess+=1
  if (isspecial_char(user_input)==True) :
    sucess+=1
  return sucess
user_input=input("Enter password : ")
strength=password_strength(user_input)
print(f"Strength points : {strength}/5")
if (strength==1) :
  print("Password Strength : VERY WEAK")
elif (strength==2) :
  print("Password Strength : LOW")
elif (strength==3) :
  print("Password Strength : MEDIUM")
elif (strength==4) :
  print("Password Strength : STRONG")
elif (strength==5) :
  print("Password Strength : VERY STRONG")
# COMMENTS!!
if (len(user_input)<6):
  print("Comment : Too short ❌ — add more characters (≥6)")
if(islower(user_input)==False) :
  print("Comment : Missing lowercase letters ❌ — add some lowercase letters")
if (isupper(user_input)==False) :
  print("Comment : Missing uppercase letters ❌ — add some uppercase letters")
if(is_digit(user_input)==False ) :
  print("Comment : Missing numbers ❌ — add at least one digit")
if(isspecial_char(user_input)==False) :
  print("Comment : Missing special characters ❌ — add at least one special character")
  print("          Special characters are : @ ,# ,$ ,! ,% ,& ,*")
else :
  print("Comment : Password is strong 🔥 — good mix of letters, numbers, and symbols")
```

**7) ROCK,PAPER,SCISSORS!!!**

```python
import random
def repeat():
  choice=input("Do you wanna continue ?(yes/no) : ").lower()
  if (choice=="yes") :
    return True
  elif (choice=="no") :
    return False
  else :
    print("Invalid choice!\nPlease re-enter your choice ")
    return repeat()
def user_choice_(options) :
  user_choice=input("\nEnter your choice : Rock,Paper,Scissors ?? : ").lower()
  if(user_choice in options ):
    return user_choice
  else :
    print("Invalid choice!\nPlease re-enter your choice ")
    return user_choice_(options)
options=["rock","paper","scissors"]
print("***\t\t\t\t\tWELCOME TO ROCK,PAPER,SCISSORS!!!\t\t\t\t\t***")
user_won=tot=comp_won=user_lost=0
continue_=True
while (continue_) :
  tot+=1
  print("\n"+"\t"*8+f"*ROUND {tot}*")
  comp_choice=random.choice(options)
  user_choice=user_choice_(options)
  print(f"Computer choose : {comp_choice}")
  if(comp_choice!=user_choice) :
    if( (comp_choice==options[0] and user_choice==options[2]) or (comp_choice==options[2] and user_choice==options[1]) or (comp_choice==options[1] and user_choice==options[0])) :
      print("\nYou lost !")
      user_lost+=1
      comp_won+=1
    else :
      print("\nYou won !!")
      user_won+=1
  elif(comp_choice==user_choice) :
    print(f"\nIt's a draw 🤝,Both chose {comp_choice}")
  print(f"\nSCORE           :  YOU - {user_won}       |  COMPUTER - {comp_won}")
  continue_=repeat()
winning_per=int(user_won*100/tot)
print("\n\n*****      GAME SUMMARY     *****")
print(f"Total rounds            : {tot} ")
print(f"Total rounds you won    : {user_won}")
print(f"Total rounds you lost   : {user_lost}")
print(f"Winning percentage      : {winning_per}%")
if(winning_per==100):
  print("UNDEFEATED 🔥")
elif(winning_per>=50) :
  print("Nice game!")
else :
  print("Comeback stronger 💪")
```

**PASSWORD GENERATOR**

```python
import random
L=list()
U=list()
Password=list()
for i in range(65,91):
    U.append(chr(i))
    L.append(chr(i+32))
D=range(0,10)
S=['!','@','#','$','&','*']
for i in [L,U,D,S]:
    Password.append(random.choice(i))
Length=int(input("Enter length of password: "))
if Length<4:
    print("Invalid length!Please enter length of password minimum 4")
else:
    n=1
    while(n<=Length-4):
       Choice=random.choice([L,U,D,S])
       Password.append(random.choice(Choice))
       n+=1
    random.shuffle(Password)
    Final=""
    for j in range (0,len(Password)):
       Final+=str(Password[j])
    print(f"Generated Password: {Final}")
```

**CALC USING RECURSION**

```python
def ask(num) :
    ch= input(f"Enter 'y' to continue calculation with {num} or 'n' to start new calculation or 'x' to exit : ").lower()
    if ch not in ['n','y','x'] :
        print("Invalid choice!!")
        return ask(num)
    return ch
def operations (a,op) :
    if op == '+' :
        b = float(input(f"Enter next number to add with {a} : "))
        return b,a+b
    elif op == '-' :
        b = float(input(f"Enter next number to subtract with {a} : "))
        return b,a-b
    elif op == '*' :
        b = float(input(f"Enter next number to multiply with {a} : "))
        return b,a*b
    else :
        b = float(input(f"Enter next number to divide with {a} : "))
        if b!=0 :
            return b,a/b
        else :
            return False,False
def calc (choice , ans =0) :
    if choice == 'x' :
        return
    elif choice == 'n' :
        a = float(input("Enter first number : "))
        op = input("Pick one operation [ +,-,/,* ] : ")
        if op not in ['+','-','*','/'] :
            print("Invalid operator! ")
            return
        b , ans = operations(a,op)
        ans=round(ans,4)
        print(f"{a} {op} {b} = {ans}")
    else :
        op = input("Pick one operation [ +,-,/,* ] : ")
        ans1=ans
        if op not in ['+','-','*','/'] :
            print("Invalid operator! ")
            return
        b ,ans  = operations(ans1,op)
        ans=round(ans,4)
        if b == ans and b== False :
            print("Error! cannot be divided by zero")
            return
        print(f"{ans1} {op} {b} = {ans}")
    choice= ask(ans)
    calc(choice,ans)
calc('n')
```

```python
import tkinter as tk
from tkinter import ttk, messagebox
import json
import os

FILE_NAME = "data.json"

class App:
    def __init__(self, root):
        self.root = root
        self.root.title("Advanced Task Manager")
        self.root.geometry("700x500")

        self.data = {"tasks": [], "notes": ""}

        self.create_ui()
        self.load_data()

    def create_ui(self):
        self.notebook = ttk.Notebook(self.root)
        self.notebook.pack(fill="both", expand=True)

        # -------- TASK TAB --------
        self.task_frame = ttk.Frame(self.notebook)
        self.notebook.add(self.task_frame, text="Tasks")

        self.task_entry = tk.Entry(self.task_frame, font=("Arial", 12))
        self.task_entry.pack(pady=10)

        self.priority = ttk.Combobox(
            self.task_frame,
            values=["Low", "Medium", "High"]
        )
        self.priority.set("Medium")
        self.priority.pack()

        btn_frame = tk.Frame(self.task_frame)
        btn_frame.pack(pady=10)

        tk.Button(btn_frame, text="Add Task", command=self.add_task).grid(row=0, column=0, padx=5)
        tk.Button(btn_frame, text="Delete Selected", command=self.delete_task).grid(row=0, column=1, padx=5)

        self.task_listbox = tk.Listbox(self.task_frame, height=15)
        self.task_listbox.pack(fill="both", expand=True, padx=10, pady=10)

        # Scrollbar
        scrollbar = tk.Scrollbar(self.task_listbox)
        scrollbar.pack(side=tk.RIGHT, fill=tk.Y)
        self.task_listbox.config(yscrollcommand=scrollbar.set)
        scrollbar.config(command=self.task_listbox.yview)

        # -------- NOTES TAB --------
        self.notes_frame = ttk.Frame(self.notebook)
        self.notebook.add(self.notes_frame, text="Notes")

        self.text_area = tk.Text(self.notes_frame, wrap="word")
        self.text_area.pack(fill="both", expand=True, padx=10, pady=10)

        tk.Button(self.notes_frame, text="Save Notes", command=self.save_notes).pack(pady=5)

        # -------- SHORTCUTS --------
        self.root.bind("<Return>", lambda e: self.add_task())
        self.root.bind("<Delete>", lambda e: self.delete_task())

    def add_task(self):
        task = self.task_entry.get()
        priority = self.priority.get()

        if not task:
            messagebox.showwarning("Warning", "Task cannot be empty!")
            return

        entry = f"[{priority}] {task}"
        self.data["tasks"].append(entry)
        self.task_listbox.insert(tk.END, entry)

        self.task_entry.delete(0, tk.END)

    def delete_task(self):
        try:
            selected = self.task_listbox.curselection()[0]
            self.task_listbox.delete(selected)
            del self.data["tasks"][selected]
        except IndexError:
            messagebox.showwarning("Warning", "No task selected!")

    def save_notes(self):
        self.data["notes"] = self.text_area.get("1.0", tk.END)
        self.save_data()
        messagebox.showinfo("Saved", "Data saved successfully!")

    def save_data(self):
        with open(FILE_NAME, "w") as f:
            json.dump(self.data, f)

    def load_data(self):
        if os.path.exists(FILE_NAME):
            with open(FILE_NAME, "r") as f:
                self.data = json.load(f)

            for task in self.data["tasks"]:
                self.task_listbox.insert(tk.END, task)

            self.text_area.insert("1.0", self.data["notes"])


if __name__ == "__main__":
    root = tk.Tk()
    app = App(root)
    root.mainloop()
```
